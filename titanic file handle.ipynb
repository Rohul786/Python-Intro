```python
import pandas as pd

# Load complete Titanic dataset directly from GitHub
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

print("Columns:", df.columns.tolist())

# (i) Clean dataset
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Fare'] = df['Fare'].fillna(df['Fare'].median())
df['Cabin'] = df['Cabin'].fillna('Unknown')
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

print("\nCleaned Dataset:")
print(df)

# (ii) Female passengers above 30 years
print("\nFemale Passengers Above 30:")
print(df[(df['Age'] > 30) & (df['Sex'] == 'female')])

# (iii) Passengers of Class 1
print("\nClass 1 Passengers:")
print(df[df['Pclass'] == 1])

# (iv) Remove duplicate Passenger IDs
df = df.drop_duplicates(subset=['PassengerId'])

print("\nAfter Removing Duplicates:")
print(df)

# (v) Insert new column
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

print("\nFinal Dataset:")
print(df)
```
