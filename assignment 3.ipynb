```python
import pandas as pd 
```

```python
df = pd.DataFrame({
    'Emp_ID': [101, 102, 103, 104, 105],
    'Emp_Name': ['Raju', 'Amit', 'Priya', 'Rahul', 'Sita'],
    'Kolkata': ['9876543210', '9876543211', None, '9876543213', None],
    'Delhi': [None, '9876543214', '9876543215', None, '9876543216'],
    'Mumbai': ['9876543217', None, None, '9876543218', '9876543219']
})

print("Wide Format:")
print(df)
```

```python
long_df = pd.melt(
    df,
    id_vars=['Emp_ID', 'Emp_Name'],
    var_name='Project_Location',
    value_name='Contact_No'
)
```

```python
long_df = long_df.dropna(subset=['Contact_No'])

print("\nLong Format:")
print(long_df.to_string(index=False))
```
