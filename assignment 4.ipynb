```python
class Bill:

    def __init__(self, consultation, medicine):
        self.consultation = consultation
        self.medicine = medicine

    def total(self):
        return self.consultation + self.medicine


bill1 = Bill(5000, 300)

print("\nBilling")
print("Consultation:", bill1.consultation)
print("Medicine:", bill1.medicine)
print("Total Bill:", bill1.total())
```

```python
import datetime

print("\nHospital Report")
print("Report Generated On:", datetime.datetime.now())
print("Patient:", patient["name"])
print("Total Bill:", bill1.total())
```

```python
# 2D array of marks
marks = [
    [80, 70, 90],
    [60, 75, 85],
    [90, 88, 95],
    [55, 65, 70],
    [78, 82, 80]
]
```

```python
# i. Maximum marks
print("Maximum marks:", max(max(row) for row in marks))

# ii. Minimum marks
print("Minimum marks:", min(min(row) for row in marks))

# iii. Average marks
total = sum(sum(row) for row in marks)
average = total / 15

print("Average marks:", average)
```

```python
# iv. Maximum marks subject-wise
print("Maximum marks subject-wise:")

for j in range(3):
    maximum = max(marks[i][j] for i in range(5))
    print("Subject", j + 1, ":", maximum)

# v. Average marks subject-wise
print("Average marks subject-wise:")

for j in range(3):
    total = sum(marks[i][j] for i in range(5))
    average = total / 5
    print("Subject", j + 1, ":", average)
```
